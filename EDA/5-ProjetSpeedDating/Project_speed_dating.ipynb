```python
# Test pour hamoniser les vagues 6 à 9. 6 à 60 si x10, 1 => 0 et 10 => 9, amplitude 10 - 1 = 9

six_att = [5, 7, 8, 1, 6, 8]
decal = [x - 1 for x in six_att]
print(decal)
rescal = [x * 100 / 9 for x in decal]
print(rescal)
somme = sum(rescal)
print(somme)
renorm = [x * 100 / somme for x in rescal]
print(renorm)
print(f"somme : {sum(renorm)}")
arrond = [round(x, 2) for x in renorm]
print(arrond)
print(f"somme arrondie : {sum(arrond)}")
```

```python
# les vagues 6 à 9 vont être harmonisées. min 6 à max 60 si x10, 1 => 0 et 10 => 9, amplitude 10 - 1 = 9

df1 = df.copy()

wav6_9 = df1["wave"].between(6, 9)
df1[wav6_9]
```

```python
# Voir quelles colonnes normaliser, pour les vagues de 6 - 9.
# A norm : attr4_1, attr3_1, attr3_s, attr3_2, attr4_2, attr4_3, attr2_3, attr3_3, attr_0, attr 
# déjà norm :pf_o_att, attr1_1, attr2_1, attr1_s, attr1_2, attr2_2, attr1_3,
# absent : attr5_1, attr7_2 , attr5_2, attr7_3, attr5_3 

# autres vagues sur 100 : pf_o_att, attr1_1, attr2_1, att1_2, attr1_3
# 1 à 10 : attr_o, attr3_1, attr, attr3_2, attr3_3

# pareil : pf_o_att, attr_o, attr_1, attr2_1, attr3_1, 

# => il n'y aurait que ces 5x6-2 colonnes à renormaliser, les autres étant notées sur la même échelle, des erreurs ds dataset des notes de 12/10 par ex... attr3_3, ligne 3130...
# A norm : attr4_1, attr3_s, attr4_2, attr4_3, attr2_3, 

verif_val = ["pf_o_att", "attr1_1", "attr4_1", 'attr2_1', 'attr3_1', 'attr5_1', "attr_o", "attr", "attr1_s", "attr3_s", "attr7_2", "attr1_2", "attr4_2", "attr2_2",
              "attr3_2", "attr5_2", "attr1_3", "attr7_3", "attr4_3", "attr2_3", "attr3_3", "attr5_3"]

# print(df1[verif_val].count(), )
# print(wav6_9[verif_val].count())
```

```python
# Renormaliser colonnes vagues 6-9

col_norm = [["attr4_1", "sinc4_1", "intel4_1", "fun4_1", "amb4_1", 'shar4_1'], 
            ["attr4_2", "sinc4_2", "intel4_2", "fun4_2", "amb4_2", "shar4_2"],
            ["attr4_3", "sinc4_3", "intel4_3", "fun4_3", "amb4_3", "shar4_3"],
            ["attr2_3", "sinc2_3", "intel2_3", "fun2_3", "amb2_3", "shar2_3"]]


for col_list in col_norm:
    for col in col_list :
        df1.loc[wav6_9, col] = (df1.loc[wav6_9, col] - 1) * 100 / 9

    for idx_ligne in df1[wav6_9].index:
        ligne_vals = df1.loc[idx_ligne, col_list]
        somme_ligne = sum(ligne_vals)
        df1.loc[idx_ligne, col_list] = ligne_vals * 100 / somme_ligne
        
for col in col_norm:
    df1.loc[wav6_9, col] = df1.loc[wav6_9, col].round(2)


df1[wav6_9]["attr3_2"]
```

```python
# Correction des valeurs 0 en 1 pour respecter l'échelle de notation de 1 à 10.

df1[["attr3_3", "sinc3_3", "intel3_3", "fun3_3", "amb3_3", "attr_o","sinc_o","intel_o","fun_o","amb_o","shar_o","attr","sinc", "intel", "fun","amb", "shar"]].describe()

col_mask_1 = ["attr3_3", "sinc3_3", "intel3_3", "fun3_3", "amb3_3", "attr_o","sinc_o","intel_o","fun_o","amb_o","shar_o","attr","sinc", "intel", "fun","amb", "shar"]
mask_1 = df1[col_mask_1] <= 0
df1[col_mask_1] = df1[col_mask_1].mask(mask_1, 1)

df1[col_mask_1].min()
```

```python
# Correction des valeurs 12, 11 et 10.5 en 10

df1[["attr3_3", "sinc3_3", "intel3_3", "fun3_3", "amb3_3", "attr_o", "fun_o"]].describe()

col_mask_12 = ["attr3_3", "sinc3_3", "intel3_3", "fun3_3", "amb3_3", "attr_o", "fun_o"]
mask_12 = df1[col_mask_12] > 10
df1[col_mask_12] = df1[col_mask_12].mask(mask_12, 10)

df1[col_mask_12].max()
```

```python
# Normalisation colonnes sur toutes les vagues

col_norm_2 = [["attr3_1", "sinc3_1", "intel3_1", "fun3_1", "amb3_1"],
              ["attr3_s", "sinc3_s", "intel3_s", "fun3_s", "amb3_s"],
              ["attr3_2", "sinc3_2", "intel3_2", "fun3_2", "amb3_2"],
              ["attr3_3", "sinc3_3", "intel3_3", "fun3_3", "amb3_3"],
              ["attr5_1", "sinc5_1", "intel5_1", "fun5_1", "amb5_1"],
              ["attr5_2", "sinc5_2", "intel5_2", "fun5_2", "amb5_2"],
              ["attr5_3", "sinc5_3", "intel5_3", "fun5_3", "amb5_3"],
              ["attr_o", "sinc_o", "intel_o", "fun_o", "amb_o", "shar_o"],
              ["attr", "sinc", "intel", "fun", "amb", "shar"]]

for col_list in col_norm_2:
    for col in col_list :
        df1.loc[:, col] = (df1.loc[:, col] - 1) * 100 / 9

    for idx_ligne in df1.index:
        ligne_vals = df1.loc[idx_ligne, col_list]
        somme_ligne = sum(ligne_vals)
        df1.loc[idx_ligne, col_list] = ligne_vals * 100 / somme_ligne
        
for col in col_norm_2:
    df1.loc[:, col] = df1.loc[:, col].round(2)
```

```python
df_moy_q1 = df_q1_long.groupby(['gender', 'temps', 'attributs'])['valeur'].mean().reset_index()
# df_moyennes
fig1 = px.line(df_moy_q1, x="temps", y="valeur", color="attributs", facet_col="gender", title="Ce que le participant recherche chez le sexe opposé ?,")
fig1.show()
```

### Pour les femmes
- La valeur la plus recherchée au temps 0 qui correspond aux attentes avant de s'inscrire à l'évènement est l'intelligence (21) suivi par la sincérité (18) et l'attractivité (18) et le fun(17), les 2 attentes les plus faibles sont les intérêts partagés (13) et l'ambition (13). 
- Au temps 1 qui correspond aux attentes à mi chemin pendant l'évènement, l'écart se réduit, toutes les variables diminuent sauf l'attractivité et les intérêtes partagés augmentent légèrement.
- Au temps 2, le lendemain de la participation à l'évènement, l'attractivié, le fun et les intérêts partagés progressent, pas de changement pour la sincérité, les attentes d'ambition diminue un peu.
- Au temps 3, 3 à 4 semaines après l'envoi des matches, les places s'affinent, l'attractivié (21) reste en tête suivi de près de l'intelligence (20), puis la sincérité (17) le fun (15), les intérêts partagés (14) et l'ambition (12).
### Pour les hommes
- t0, c'est l'attractivié qui prédomine (27), suivi par par l'intelligence (20), le fun (18), la sincérité (16), les intérêts partagés (11) et l'ambition (9);
- t1, même classement, juste les valeurs se recentrent, (24-18-15-14-11-10).
- t2, l'attractivité remonte en flèche (30), le fun (18) change sa place avec l'intelligence (17), les autres restent égales (15-12-8).
- t3, l'attractivité reste en tête (29), l'intelligence reprend sa place intiale (19), pour le reste rien ne change (17-16-1-9).
### Conclusion
L'écart est plus étalé chez les hommes que chez les femmes. Les variables intelligence, fun et sincérité se placent également considérés chez les 2 groupes. L'attractivité est plus importance pour les hommes, les intérêts partagés et l'ambition sont moins considérés des 2 côtés. Les attentes sont à peu près les mêmes, seuls les valeurs divergent.

## Pour obtenir un deuxième rendez-vous, vaut-il mieux être le premier ou le dernier rendez-vous de la soirée ?

```python
premier_rdv = df1[df1["order"] == 1].copy()
# premier_rdv
dernier_rdv = df1.groupby("iid")["order"].max().reset_index()
dernier_rdv.columns = ["iid", "max_order"]
# dernier_rdv

df1_avec_max = df.merge(dernier_rdv, on="iid")
# df1_avec_max

dernier_rdv = df1_avec_max[df1_avec_max['order'] == df1_avec_max['max_order']].copy()
# dernier_rdv

print("=== RÉSULTATS ===")
print(f"Nombre de premiers rdv : {len(premier_rdv)}")
print(f"Nombre de derniers rdv : {len(dernier_rdv)}")

taux_premier = premier_rdv['dec'].mean()
taux_dernier = dernier_rdv['dec'].mean()

print(f"\nTaux de succès premier rdv : {taux_premier:.3f} ({taux_premier*100:.1f}%)")
print(f"Taux de succès dernier rdv : {taux_dernier:.3f} ({taux_dernier*100:.1f}%)")

difference = taux_dernier - taux_premier
print(f"\nDifférence : {difference:.3f} points")

if difference > 0:
    print("Le dernier rendez-vous semble plus favorable")
elif difference < 0:
    print("Le premier rendez-vous semble plus favorable") 
else:
    print("Pas de différence notable")
```
