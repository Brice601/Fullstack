# Fiche récapitulative : Machine Learning - Régression et Régularisation

## 1. Notations de base

### Symboles mathématiques
- **X^t** : X transposée (pas une puissance !) - échange lignes et colonnes
- **|| ||₂** : norme euclidienne (√(x₁² + x₂² + ...))
- **|| ||₁** : norme L1 (|x₁| + |x₂| + ...)
- **|| ||₀** : norme L0 (nombre d'éléments non-nuls)
- **∀** : "pour tout" (quantificateur universel)
- **R₊*** : nombres réels strictement positifs (> 0, excluant zéro)
- **argmin** : "l'argument qui minimise"

### Structure des données
- **DataFrame → Matrice** :
  - Chaque **ligne** du DataFrame = un **vecteur** dans la matrice X
  - Chaque **colonne** du DataFrame = une **variable/feature**
  - **X** : matrice des variables explicatives (n × p)
  - **y** : vecteur de la variable cible (n × 1)

## 2. Fonction de coût et optimisation

### Fonction de coût classique
```
||y - X^t β||²₂
```
**Signification** : "Distance au carré entre vraies valeurs et prédictions"

**Se lit** : "Norme deux au carré de y moins X transposée fois bêta"

### Optimisation
```
β* = argmin_β(||y - X^t β||²₂)
```
**Signification** : "Trouve le β qui minimise l'erreur"

**Comment ça marche** :
- L'ordinateur utilise des algorithmes d'optimisation (descente de gradient)
- Commence avec des valeurs aléatoires
- Ajuste β itérativement pour réduire l'erreur
- S'arrête quand l'erreur ne peut plus diminuer

## 3. Bias vs Variance

### Décomposition de l'erreur
```
Erreur totale = Bias² + Variance + σ²
```

### Analogie du tireur à l'arc
- **Bias** : Est-ce que je vise bien le centre en moyenne ?
  - Bias élevé = toujours décalé dans la même direction
  - Bias faible = en moyenne, je vise le centre

- **Variance** : Est-ce que mes tirs sont réguliers ?
  - Variance élevée = tirs éparpillés
  - Variance faible = tirs groupés

- **Bruit σ²** : Part d'erreur irréductible (aléatoire)

### Trade-off
- **Modèles simples** : Bias élevé, Variance faible
- **Modèles complexes** : Bias faible, Variance élevée
- **Objectif** : Trouver l'équilibre optimal

## 4. Le problème p > n

### Situation problématique
- **p** = nombre de variables
- **n** = nombre d'observations
- Si **p > n** : plus de paramètres à estimer que de données

### Conséquences
- Système d'équations **sous-déterminé**
- **Infinité de solutions** possibles
- Matrice **X^t X non-invertible**
- **Impossible d'estimer β** de façon unique

### Solution : La régularisation !

## 5. Régularisation

### Ridge Regression
```
||y - X^t β||²₂ + λ||β||²₂
```

**Caractéristiques** :
- **λ** contrôle la régularisation (λ ∈ R₊*)
- **Réduit** tous les coefficients vers zéro
- **Ne supprime jamais** de variables (β ≠ 0)
- **Rend la matrice invertible** : (X^t X + λI)

**Effet du paramètre λ** :
- λ = 0 → régression linéaire classique (variance max, bias min)
- λ augmente → bias augmente, variance diminue
- λ = ∞ → tous β = 0, modèle prédit juste la moyenne

### Lasso Regression
```
||y - X^t β||²₂ + λ||β||₁
```

**Caractéristiques** :
- Utilise la **norme L1** (somme des valeurs absolues)
- **Sélection automatique** de variables
- Met certains **β exactement à zéro**
- **Régularisation + sélection** en une seule étape

### Best Subset Selection
```
β̂ = argmin_{β: ||β||₀ ≤ s} ||Y - X^t β||²₂
```

**Caractéristiques** :
- Sélectionne **exactement s variables**
- Toutes les autres ont β = 0
- **Très difficile** à calculer (problème combinatoire)
- Lasso est une approximation plus facile

## 6. Comparaison des méthodes

| Méthode | Sélection variables | Coefficients | Calcul | Usage |
|---------|-------------------|--------------|--------|--------|
| **Ridge** | Non | Tous réduits vers 0 | Facile | Quand toutes variables utiles |
| **Lasso** | Oui, automatique | Certains = 0 exactement | Facile | Quand beaucoup de bruit |
| **Best Subset** | Oui, contrôlée | s non-nuls, autres = 0 | Difficile | Quand on sait combien garder |

## 7. Concepts clés à retenir

### Optimisation
- Les algorithmes trouvent β* automatiquement
- Pas besoin de tester "toutes" les valeurs
- Processus itératif intelligent (descente de gradient)

### Régularisation
- Ajoute une contrainte/pénalité au modèle
- Permet de résoudre le problème p > n
- Contrôle le trade-off bias-variance

### Sélection de variables
- Ridge : garde tout, réduit tout
- Lasso : sélectionne automatiquement
- Best Subset : contrôle exact du nombre

### Le paramètre λ
- **Hyperparamètre** à optimiser
- Plus λ est grand, plus le modèle est "discipliné"
- À choisir par validation croisée
